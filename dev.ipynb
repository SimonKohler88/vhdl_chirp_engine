```python
def calc_sine_table(addr_range, amplitude_bitwidth, amplitude_bitwidth_name='C_AMP_WIDTH'):
    ampl = (2 ** amplitude_bitwidth - 2)/2 # todo range of signed
    s = ''
    for i in range(addr_range):
        phase = 2* np.pi / addr_range * i
        # 62 => to_signed( -6393,16),
        s += f'\t\t{i:<5} => to_signed({round(ampl* np.sin(phase)):>7}, {amplitude_bitwidth_name})'
        if i != addr_range - 1:
            s += ',\n'
        else:
            s += '\n'
    print(s)

def calc_quarter_sine_table(addr_range, amplitude_bitwidth, amplitude_bitwidth_name='C_AMP_WIDTH'):
    ampl = (2 ** amplitude_bitwidth - 2)/2 # todo range of signed
    s = ''
    for i in range(addr_range):
        phase = 0.5* np.pi / addr_range * i
        # 62 => to_signed( -6393,16),
        s += f'\t\t{i:<5} => to_signed({round(ampl* np.sin(phase)):>7}, {amplitude_bitwidth_name})'
        if i != addr_range - 1:
            s += ',\n'
        else:
            s += '\n'
    print(s)
calc_quarter_sine_table(1024, 16)
```
